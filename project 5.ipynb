```python
df = pd.read_excel("States_Annual_Energy_Generation_Sources_1990_2019.xlsx", header=1)
df = df[df['TYPE OF PRODUCER']=='Total Electric Power Industry']
```

```python
yearly = df.groupby(['YEAR','ENERGY SOURCE'])['GENERATION (Megawatthours)'].sum().reset_index()
pivot_year = yearly.pivot(index='YEAR', columns='ENERGY SOURCE', values='GENERATION (Megawatthours)').fillna(0)
```

```python
renewable_cols = ['Wind','Solar Thermal and Photovoltaic','Hydroelectric Conventional','Geothermal','Wood and Wood Derived Fuels']
fossil_cols = ['Coal','Natural Gas','Petroleum']
```

```python
pivot_year['Renewable_Total'] = pivot_year[[c for c in renewable_cols if c in pivot_year.columns]].sum(axis=1)
pivot_year['Fossil_Total'] = pivot_year[[c for c in fossil_cols if c in pivot_year.columns]].sum(axis=1)
pivot_year['Total'] = pivot_year['Renewable_Total'] + pivot_year['Fossil_Total']
pivot_year['Renewable_Pct'] = pivot_year['Renewable_Total'] / pivot_year['Total'] * 100
```

```python
print(pivot_year[['Renewable_Total','Fossil_Total','Renewable_Pct']].tail())
```

```python
X = pivot_year.index.values.reshape(-1,1)  # Year
y = pivot_year['Renewable_Pct'].values
```

```python
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)
```

```python
future_years = np.arange(2020, 2031).reshape(-1,1)
future_pred = model.predict(future_years)
```

```python
plt.figure(figsize=(8,5))
plt.plot(pivot_year.index, y, marker='o', label='Actual Renewable %')
plt.plot(pivot_year.index, y_pred, linestyle='--', label='Model Fit')
plt.plot(future_years, future_pred, marker='x', color='red', linestyle='--', label='Forecast 2020-2030')
plt.xlabel('Year')
plt.ylabel('Renewable %')
plt.title('TASK 6.1: Forecast of Clean Energy Production')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
final_energy = pd.read_csv("final_cleaned_data.csv")
final_energy['STATE'] = final_energy['STATE'].str.strip()
final_energy['EV_PERCENT'] = final_energy['EV_REGISTRATIONS'] / final_energy['GRAND_TOTAL'] * 100
```

```python
df_e = pd.read_excel("States_Annual_Energy_Generation_Sources_1990_2019.xlsx", header=1)
df_e = df_e[(df_e['YEAR']==2018) & (df_e['TYPE OF PRODUCER']=='Total Electric Power Industry')]
df_e['STATE'] = df_e['STATE'].astype(str).str.strip()
pivot_e = df_e.pivot_table(index='STATE', columns='ENERGY SOURCE', values='GENERATION (Megawatthours)', aggfunc='sum').reset_index()
pivot_e = pivot_e.fillna(0)
```

```python
full_df = pd.merge(final_energy, pivot_e, on='STATE', how='left')
full_df = full_df.fillna(0)
full_df.to_csv("full_final_energy_ev.csv", index=False)  # Ab save ho jayega, agli baar error nahi
print("full_df ban gaya:", full_df.shape)
print(full_df[['STATE','EV_PERCENT','Wind','Solar Thermal and Photovoltaic']].head())
```

```python
cols = ['Wind','Solar Thermal and Photovoltaic','Hydroelectric Conventional']
X2 = full_df[cols]
y2 = full_df['EV_PERCENT']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X2, y2, test_size=0.3, random_state=42)
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
pred = rf.predict(X_test)
```

```python
r2 = r2_score(y_test, pred)
rmse = np.sqrt(mean_squared_error(y_test, pred))
print(f"\n=== TASK 6.2 METRICS ===")
print(f"R2 Score: {r2:.3f}")
print(f"RMSE: {rmse:.4f}")
```
